```python
#Instalación de dependencias
import time
import torch

import pandas as pd

from datasets import load_dataset
from getpass import getpass
from huggingface_hub import login
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from transformers import pipeline
```

```python
#Configuración de GPU
device = 0 if torch.cuda.is_available() else -1

if device == 0:
    print("GPU disponible:", torch.cuda.get_device_name(0))
else:
    print("GPU no disponible. Se utilizará CPU.")
```

```python
#Autenticación en Hugging Face
token = getpass("Ingresa tu token de Hugging Face: ")
login(token = token)

print("Autenticación completada.\n")
```

```python
#Carga del dataset
dataset = load_dataset(
    "nyu-mll/glue",
    "sst2",
    split = "validation"
)
```

```python
#Usar los mismos 200 registros que con BERT Tiny
dataset = dataset.select(range(200))

print(dataset)
print(dataset[0])
```

```python
#Carga del modelo
modelo = "distilbert/distilbert-base-uncased-finetuned-sst-2-english"

clasificador = pipeline(
    "text-classification",
    model = modelo,
    framework = "pt",
    device = device
)

print("\nModelo cargado correctamente.\n")
```

```python
#Inferencia y medición de latencia
textos = list(dataset["sentence"])
etiquetas_reales = list(dataset["label"])

inicio = time.time()

resultados = clasificador(
    textos,
    batch_size = 8
)

fin = time.time()

tiempo_total = fin - inicio
latencia = tiempo_total / len(textos)

print("Tiempo total:", round(tiempo_total, 2), "segundos")
print("Latencia promedio:", round(latencia, 4), "segundos por texto\n")
```

```python
#Conversión de predicciones
predicciones = []

for resultado in resultados:
    if resultado["label"] == "POSITIVE":
        predicciones.append(1)
    else:
        predicciones.append(0)
```

```python
#Cálculo de métricas
accuracy = accuracy_score(etiquetas_reales, predicciones)

precision = precision_score(
    etiquetas_reales,
    predicciones
)

recall = recall_score(
    etiquetas_reales,
    predicciones
)

f1 = f1_score(
    etiquetas_reales,
    predicciones
)

print("Accuracy:", round(accuracy, 4))
print("Precision:", round(precision, 4))
print("Recall:", round(recall, 4))
print("F1-score:", round(f1, 4), "\n")
```

```python
#Tabla resumen
resumen = pd.DataFrame({
    "Modelo": ["DistilBERT"],
    "Registros": [len(dataset)],
    "Accuracy": [accuracy],
    "Precision": [precision],
    "Recall": [recall],
    "F1-score": [f1],
    "Latencia_seg": [latencia]
})

print(resumen)

resumen.to_csv(
    "resultados_distilbert.csv",
    index = False
)
```
